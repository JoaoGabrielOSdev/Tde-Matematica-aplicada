## Etapa 2 – Identificar as variáveis

As duas variáveis presentes na base são:

- `mes`
- `custo_transporte`

### a) Variável independente

A variável independente é o **mês**, pois representa a passagem do tempo e é utilizada para analisar a variação dos custos.

### b) Variável dependente

A variável dependente é o **custo de transporte**, pois seu valor é analisado em função do mês.

### c) Representação das variáveis

- **Mês:** representa o período de tempo analisado, variando de 1 a 12.
- **Custo de transporte:** representa o custo total de transporte da empresa em cada mês, em reais.

```python
#Construa um gráfico de dispersão utilizando os dados da base.
plt.scatter(
    df["mes"],
    df["custo_transporte"],
    label="Dados observados")

plt.title("Custos de Transporte por Mês")
plt.xlabel("Mês")
plt.ylabel("Custo de Transporte (R$)")
plt.xticks(range(1, 13))
plt.grid(True, alpha=0.3)
plt.legend()

plt.show()
```

O gráfico mostra a o aumento do custo com o passar dos meses, apesar de pequenas variações em alguns momentos, o comportamento geral é cresimento.

```python
# Variável independente
x = df["mes"]

# Variável dependente
y = df["custo_transporte"]

# Encontrar os coeficientes da função afim
a, b = np.polyfit(x, y, 1)

print(f"Coeficiente a = {a:.2f}")
print(f"Coeficiente b = {b:.2f}")

print(f"\nFunção encontrada:")
print(f"f(x) = {a:.2f}x + {b:.2f}")
```

## Etapa 5 – Interpretar a função

A função encontrada foi:

**f(x) = 1791,61x + 42.204,55**

### a) O que representa o coeficiente a?

O coeficiente **a = 1.791,61** representa a variação aproximada do custo de transporte a cada mês.

Isso significa que, de acordo com a tendência representada pela função, o custo de transporte aumenta aproximadamente **R$ 1.791,61 por mês**.

### b) O que representa o coeficiente b?

O coeficiente **b = 42.204,55** representa o valor estimado pela função quando x = 0.

Embora o mês 0 não esteja presente na base analisada, esse valor corresponde ao ponto em que a reta intercepta o eixo dos custos.

### c) O valor de a indica crescimento ou redução?

Como o coeficiente **a é positivo**, ele indica uma tendência de **crescimento dos custos de transporte ao longo do tempo**.

```python
# Criar os valores da reta
x_reta = np.linspace(df["mes"].min(), df["mes"].max(), 100)
y_reta = a * x_reta + b

plt.figure(figsize=(10, 6))

# Dados reais
plt.scatter(
    df["mes"],
    df["custo_transporte"],
    label="Dados observados"
)

# Reta da função
plt.plot(
    x_reta,
    y_reta,
    label=f"f(x) = {a:.2f}x + {b:.2f}"
)

plt.title("Custos de Transporte e Função Afim")
plt.xlabel("Mês")
plt.ylabel("Custo de Transporte (R$)")
plt.grid(True, alpha=0.3)
plt.xticks(range(1, 13))
plt.legend()

plt.show()
```

```python
# Meses que serão previstos
meses_futuros = np.array([13, 14, 15])

# Calcular as previsões
previsoes = a * meses_futuros + b

# Exibir os resultados
for mes, custo in zip(meses_futuros, previsoes):
    print(f"Mês {mes}: R$ {custo:,.2f}")
```

## Etapa 8 – Análise dos resultados

### a) A função encontrada representa bem a tendência dos dados?

Sim. A função representa de forma aproximada a tendência geral dos dados. Os valores reais apresentam pequenas oscilações em relação à reta, mas o comportamento geral é crescente.

### b) Os custos estão aumentando ou diminuindo?

Os custos de transporte estão **aumentando ao longo dos meses**. Os dados começam em R$ 44.500,00 no primeiro mês e chegam aR$ 64.000,00 no décimo segundo mês.

De acordo com o coeficiente angular da função, o custo aumenta aproximadamente **R$ 1.791,61 por mês**.
